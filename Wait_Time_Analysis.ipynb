# 🏠 UBC Student Housing Wait Time Analysis

## 📌 Problem Statement

Year-round housing at UBC is highly competitive, particularly among **international and graduate students**. While first-year students often receive a guarantee, others face long, uncertain waits. For example:

- **Shared rooms**, one of the most affordable options, can involve **wait times between 1.5 and 2 years**
-  UBC advises returning students to apply “as soon as possible,” as waitlists *may take up to 1.5 years* to clear ([source](https://learningcommons.ubc.ca/residence-planning-ahead-for-year-round-housing/))
- Wait times and waitlist movement **vary significantly by room type**, and UBC *does not* publicly provide personalized guidance on this

As a result, students are left guessing when to apply and which housing type to pursue.

## 🎯 Project Objective

This project aims to empower UBC students (especially internationals) with personalized, data-driven insights by:  
1. Calculating the **average wait time for each room type** (e.g., Shared, Studio, 1BR)  
2. Assessing whether **application month** significantly affects wait times  
3. Comparing **wait times across student categories** (e.g., international vs domestic, undergraduate vs graduate)  
4. Identifying which **housing types are most oversubscribed**, indicating demand-supply mismatches  
5. Forecasting **wait times into 2025–2026** to help future applicants plan  
6. Providing **actionable recommendations**—including application timing and room type strategies—to reduce wait time and improve housing outcomes


By the end, students will no longer have to guess—they can make **informed decisions** based on data-driven insights.
